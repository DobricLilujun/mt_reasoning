```python
df = pd.DataFrame()
for f in glob.glob("data/extraction_pdf/datasets/SUMMARY/*"):
    if "results_summary_task" in f:
        if "task4" in f:
            df_task4 = pd.read_json(f, lines=True)

def split_model(input_str):
    model = input_str.split("_")[3].split(".jsonl")[0]
    return model

df_task4["model"] = df_task4.file_name.apply(split_model)
models_all = ["gpt-5","Llama-3.3-70B-Instruct", "Qwen3-Next-80B-A3B-Instruct","gemma-3-27b-it","Qwen3-30B-A3B-Instruct-2507","gemma-2-2b-it","gemma-2-9b-it", "gemma-3-4b-it","gemma-3-12b-it","Llama-3.2-1B-Instruct","Llama-3.2-3B-Instruct","Qwen3-4B-Instruct-2507"]
df_task4 = df_task4[df_task4["model"].isin(models_all)]
df_task4 = df_task4[["model", "accuracy"]]

df_task4.rename(columns={"accuracy": "task4"}, inplace=True)
df_task_1_2_3 = df_task_1_2_3[["model", "task","accuracy"]]

df_flat = df_task_1_2_3.pivot(index='model', columns='task', values='accuracy').reset_index()
df_task_1_2_3_4 = pd.merge(df_flat, df_task4, on="model", how="left")
```

```python
df_summary = pd.merge(df_task_1_2_3_4, df_trans, on="model", how="left")

# 归一化 CometScore
c = df_summary['CometScore'].astype(float)
c_min, c_max = c.min(), c.max()
df_summary['CometScore'] = (c - c_min) / (c_max - c_min + 1e-12)

# 归一化 CometScoreSmall
c = df_summary['CometScoreSmall'].astype(float)
c_min, c_max = c.min(), c.max()
df_summary['CometScoreSmall'] = (c - c_min) / (c_max - c_min + 1e-12)

# 归一化 spbleu_score
c = df_summary['spbleu_score'].astype(float)
c_min, c_max = c.min(), c.max()
df_summary['spbleu_score'] = (c - c_min) / (c_max - c_min + 1e-12)  

# 归一化 chrf_score
c = df_summary['chrf_score'].astype(float)
c_min, c_max = c.min(), c.max()
df_summary['chrf_score'] = (c - c_min) / (c_max - c_min + 1e-12)


```

```python


cols = ['task1', 'task2','task3', 'task4', 'CometScore', 'CometScoreSmall', 'spbleu_score', 'chrf_score']
num_df = df_summary[cols].select_dtypes(include=['number'])

corr_p = num_df.corr(method='pearson')
corr_s = num_df.corr(method='spearman')

fig, axes = plt.subplots(nrows=2, figsize=(8, 10), constrained_layout=True)

rename_map = {
    'spbleu_score': 'spBLEU',
    'chrf_score': 'Chrf++',
    'CometScore': 'Comet',
    'CometScoreSmall': 'Comet-Small',
    'task1': 'Task 1',
    'task2': 'Task 2',
    'task3': 'Task 3',
    'task4': 'Task 4',
}

def relabel_axis(ax, rename_map):
    # X 轴
    xt = [t.get_text() for t in ax.get_xticklabels()]
    xt_new = [rename_map.get(t, t) for t in xt]
    # 保留原有旋转与样式
    rot = ax.get_xticklabels()[0].get_rotation() if ax.get_xticklabels() else 0
    ax.set_xticklabels(xt_new, rotation=rot)
    for tick in ax.get_xticklabels():
        tick.set_fontweight('bold')
    # Y 轴（如需要同步）
    yt = [t.get_text() for t in ax.get_yticklabels()]
    yt_new = [rename_map.get(t, t) for t in yt]
    ax.set_yticklabels(yt_new)

# Pearson
ax0 = axes[0]
sns.heatmap(
    corr_p, annot=True, fmt='.2f', cmap='coolwarm',
    vmin=0.0, 
    vmax=1.0,  
    square=True, linewidths=0.5,
    cbar_kws={'label': r'Pearson \(r\)'},
    ax=ax0
)
ax0.tick_params(axis='x', labelrotation=45, labelbottom=False)
relabel_axis(ax0, rename_map)

# Spearman
ax1 = axes[1]
sns.heatmap(
    corr_s, annot=True, fmt='.2f', cmap='coolwarm',
    vmin=0.0, 
    vmax=1.0,  
    square=True, linewidths=0.5,
    cbar_kws={'label': r'Spearman \(\rho\)'},
    ax=ax1
)
ax1.tick_params(axis='x', labelrotation=45)
relabel_axis(ax1, rename_map)

# 间距更紧
fig.set_constrained_layout_pads(hspace=0.01)
plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```
