```python
import pandas as pd
import numpy as np
import glob
import os
import pandas as pd

base_dir = "data/performance_eval"
all_jsonl = glob.glob(os.path.join(base_dir, "**", "*.jsonl"), recursive=True) 
task_files = [p for p in all_jsonl if "task_eval_translation" in os.path.basename(p) or "task_eval_translation" in p]  
df_summary = pd.DataFrame()
for f in task_files:
    df = pd.read_json(f, lines=True)
    print(f"Processing file: {f} with {len(df)} records")
    mean_comet = df['CometScore'].mean()
    std_comet = df['CometScore'].std()
    mean_spbleu = df['spbleu_score'].mean()
    std_spbleu = df['spbleu_score'].std()
    mean_chrf = df['chrf_score'].mean()
    std_chrf = df['chrf_score'].std()
    print(f"CometScore: {mean_comet:.2f} ± {std_comet:.2f}")
    print(f"spbleu_score: {mean_spbleu:.2f} ± {std_spbleu:.2f}")
    print(f"chrf_score: {mean_chrf:.2f} ± {std_chrf:.2f}")
    
    row = {
        "file_name": os.path.basename(f),
        "mean_comet": mean_comet,
        "std_comet": std_comet,
        "mean_spbleu": mean_spbleu,
        "std_spbleu": std_spbleu,
        "mean_chrf": mean_chrf,
        "std_chrf": std_chrf
    }
    row_df = pd.DataFrame([row])
    df_summary = pd.concat([df_summary, row_df], ignore_index=True)

df_summary.to_json("data/extraction_pdf/datasets/SUMMARY/results_summary_translation.json", orient="records", lines=True)
```
