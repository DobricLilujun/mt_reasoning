## Evalulation of Accuracy With Logprob

### Task1

```python
import pandas as pd
import numpy as np

df = pd.read_json("data/extraction_pdf/datasets/output_gammar_results/gemma-2-2b-it/task1_20251003_202254_2_gemma-2-2b-it.jsonl", lines=True)

num_correct = 0
total = len(df)
df_final = pd.DataFrame()
for idx, row in df.iterrows():
    # print(f"Processing row {idx}")
    log_prob_dict = row["log_probs"]["content"]
    correct_grammar_letter = row["correct_grammar_letter"]

    ## Calculate NLL with Log Probabilities
    for item in log_prob_dict:
        # print(f"Decoded token: {item['token']}, Number: {item['bytes']}")
        if item['token'] == correct_grammar_letter:
            logprob_list = item["top_logprobs"]
            for log_prob_dict in logprob_list:
                if correct_grammar_letter == log_prob_dict["token"]:
                    log_probs = log_prob_dict["logprob"]
                    nll = -log_probs
                    probs_from_logprobs = np.exp(log_probs)
                    # print(f"Log probabilities for token '{correct_grammar_letter}': {log_probs}")
                    # print(f"Probabilities for token '{correct_grammar_letter}': {probs_from_logprobs}")
                    row["log_probs_value"] = log_probs
                    row["nll"] = nll
                    row["probs_from_logprobs"] = probs_from_logprobs
                    # print(f"Probabilities for token '{correct_grammar_letter}': {probs_from_logprobs}")
                    if np.isnan(nll):
                        print(f"NaN found in NLL for row {idx}")
                    break

    ## Calculate Accuracy
    correct_grammar_letter = row['correct_grammar_letter']
    detected_grammar_letter = row['task1_dict'].get('grammar_selected', '').strip().upper()
    if correct_grammar_letter == detected_grammar_letter:
        num_correct += 1
        row['is_correct'] = True
    
    else:
        row['is_correct'] = False
    
    row_df = pd.DataFrame([row])
    df_final = pd.concat([df_final, row_df], ignore_index=True)

acc = num_correct / total
std = np.sqrt(acc * (1 - acc) / total)
probs_list = df_final["probs_from_logprobs"].dropna().values
probs_mean = np.mean(probs_list)
probs_std = np.std(probs_list)

print(f"Accuracy: {num_correct}/{total} = {acc:.2f} ± {std:.2f}")
print(f"Probabilities: {probs_mean:.2f} ± {probs_std:.2f}")
```

```python
import pandas as pd
import numpy as np

df = pd.read_json("data/extraction_pdf/datasets/output_gammar_results/gemma-2-2b-it/task3_20251003_221009_4_2_1000_gemma-2-2b-it.jsonl", lines=True)

total_score = 0 
total = len(df)
df_final = pd.DataFrame()
for idx, row in df.iterrows():
    # print(f"Processing row {idx}")
    log_prob_dict = row["log_probs"]["content"]
    correct_grammar_letters = row["correct_grammars_letter"]

    ## Calculate NLL with Log Probabilities
    prob_accu = 0.0

    for correct_grammar_letter in correct_grammar_letters:
        # Find all log prob outputs
        for item in log_prob_dict:
            if item['token'] == str(correct_grammar_letter):
                logprob_list = item["top_logprobs"]
                # Find the log prob for the correct grammar letter
                for log_prob in logprob_list:
                    if str(correct_grammar_letter) == log_prob["token"]:
                        log_probs = log_prob["logprob"]
                        nll = -log_probs
                        probs_from_logprobs = np.exp(log_probs)
                        # print(f"Log probabilities for token '{correct_grammar_letter}': {log_probs}")
                        # print(f"Probabilities for token '{correct_grammar_letter}': {probs_from_logprobs}")

                        prob_accu += probs_from_logprobs
                        if np.isnan(nll):
                            print(f"NaN found in NLL for row {idx}")
                        break
                break
    row["prob_accu"] = prob_accu/len(correct_grammar_letters) if correct_grammar_letters else 2
    ## Calculate Accuracy
    correct_set = set(row['correct_grammars_letter'])
    detected_set = set(row["task3_dict"].get('grammar_selected', ''))
    score = len(detected_set & correct_set) / len(correct_set)
    total_score += score

    row["total_score"] = score
    row_df = pd.DataFrame([row])
    df_final = pd.concat([df_final, row_df], ignore_index=True)

acc = total_score / total
std = np.sqrt(acc * (1 - acc) / total)

probs_list = df_final["prob_accu"].dropna().values
probs_mean = np.mean(probs_list)
probs_std = np.std(probs_list)

print(f"Accuracy: {total_score}/{total} = {acc:.2f} ± {std:.2f}")
print(f"Probabilities: {probs_mean:.2f} ± {probs_std:.2f}")
```
