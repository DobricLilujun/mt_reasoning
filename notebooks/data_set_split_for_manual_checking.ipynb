```python
import pandas as pd

df = pd.read_json("data/extraction_pdf/datasets/df_samples.jsonl", lines=True)
```

```python
sampled = df.sample(100, random_state=42)
sampled["is_luxembourgish_correct"] = None
sampled["is_grammar_in_luxembourgish_sentence"] = None
sampled.to_json("data/extraction_pdf/datasets/df_samples_for_manual_checking_100.jsonl", lines=True, orient="records")
```

```python
sampled.to_excel("data/extraction_pdf/datasets/df_samples_for_manual_checking_100.xlsx", index=False)
```
